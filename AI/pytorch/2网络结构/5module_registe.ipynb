```python
import torch
import torch.nn as nn
import time

class MySequential(nn.Module):
    def __init__(self, *args):
        super().__init__()
        for idx, module in enumerate(args):
            # 这里，module是Module子类的一个实例。我们把它保存在'Module'类的成员
            # 变量_modules中。_module的类型是OrderedDict
            self._modules[str(idx)] = module

    def forward(self, X):
        # OrderedDict保证了按照成员添加的顺序遍历它们
        for block in self._modules.values():
            X = block(X)
        return X

# 测试推理时间
net = MySequential(nn.Linear(20, 256), nn.ReLU(), nn.Linear(256, 10))
print(list(net.parameters()))  # 子模块被正确注册和跟踪。
input_tensor = torch.randn(4, 20)

start_time = time.time()
output = net(input_tensor)
end_time = time.time()
print(f"ModelWithDict推理时间: {end_time - start_time:.6f}秒")
```

```python
import torch
import torch.nn as nn
import time

class MySequential(nn.Module):
    def __init__(self, *args):
        super().__init__()
        self.blocks = []  # 使用普通列表存储模块
        for module in args:
            self.blocks.append(module)

    def forward(self, X):
        for block in self.blocks:
            X = block(X)
        return X

# 测试推理时间
net = MySequential(nn.Linear(20, 256), nn.ReLU(), nn.Linear(256, 10))
print(list(net.parameters()))  # 输出为空，因为子模块未被追踪，所以遇到模型参数更新、模型参数保存、模型参数移动到设备(cuda)时会报错
input_tensor = torch.randn(4, 20)

start_time = time.time()
output = net(input_tensor)
end_time = time.time()
print(f"ModelWithDict推理时间: {end_time - start_time:.6f}秒")
```
