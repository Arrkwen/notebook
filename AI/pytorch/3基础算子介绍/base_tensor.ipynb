```python
import torch


x1 = torch.arange(12).reshape(3,4)
x2 = torch.arange(12).reshape((3,4))
print(x1)
print(x2)
```

```python
# 广播，x1复制列，x2复制行
x1 = torch.arange(3).reshape(3,1)
x2 = torch.arange(2).reshape(1,2)
print(x1)
print(x2)
print(x1+x2)
```

```python
# 广播，x1复制列，x2复制行
x1 = torch.arange(6).reshape(3,2)
x2 = torch.arange(4).reshape(1,4)
print(x1)
print(x2)
# print(x1+x2) # error 广播机制也不是任何情况都能广播
```

```python
# 广播，x1复制列，x2复制行
x1 = torch.arange(6).reshape(1,2,3)
x2 = torch.arange(3).reshape(1,1,3)
print(x1)
print(x2)
print(x1+x2)
```

```python
x1 = torch.arange(6).reshape(2,3)
print(x1)
print(x1.sum(axis=0)) # [1,3]
print(x1.sum(axis=1)) # [1,2]
```

```python
x1 = torch.arange(24).reshape(2,3,4)
print(x1)
print(len(x1))
print("-----------------------------")
print(x1.sum(axis=0)) # [3,4]
print(x1.sum(axis=1)) # [2.4]
print(x1.sum(axis=2)) # [2,3]
```

```python

x1 = torch.arange(20).reshape(5,4)
x2 = torch.arange(4)
print(x1)
print(x2)

print(torch.mv(x1,x2))
```

```python

x1 = torch.arange(20).reshape(5,4)
x2 = torch.arange(12).reshape(4,3)

print(x1)
print(x2)

print(torch.mm(x1,x2))
```

```python

x1 = torch.arange(4).reshape(2,2)
x2 = torch.arange(4).reshape(2,2)

print(x1)
print(x2)

print(torch.mul(x1,x2))
print(x1*x2)

```

```python
x = torch.arange(6, dtype=torch.float32).reshape(2,3)


print(x)
print(torch.linalg.norm(x))
print(torch.sqrt(torch.mul(x,x).sum()))
```
